```python
from sklearn.linear_model import Ridge
import numpy as np 
import pandas as pd 
from sklearn.model_selection import GridSearchCV, train_test_split
```

```python
df=pd.read_csv('truedata.csv')
```

```python
df = df.dropna()
```

```python
TargetVariable=['True Stress']
Predictors=['Strain Rate','Temperature','True Plastic Strain']
 
X=df[Predictors].values
y=df[TargetVariable].values
```

```python
from sklearn.preprocessing import StandardScaler
PredictorScaler=StandardScaler()
TargetVarScaler=StandardScaler()
 
# Storing the fit object for later reference
PredictorScalerFit=PredictorScaler.fit(X)
TargetVarScalerFit=TargetVarScaler.fit(y)
 
# Generating the standardized values of X and y
X=PredictorScalerFit.transform(X)
y=TargetVarScalerFit.transform(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test  = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
ridge_reg = Ridge()
params_Ridge = {'alpha': [10,1,0.1,0.01,0.001,0.0001,0.00001,0] , "fit_intercept": [True, False], "solver": ['svd', 'cholesky', 'lsqr', 'sparse_cg', 'sag', 'saga']}
Ridge_GS = GridSearchCV(ridge_reg, param_grid=params_Ridge, n_jobs=-1)
Ridge_GS.fit(X_train,y_train)
Ridge_GS.best_params_
```

```python
Ridge_GS.best_estimator_
```

```python
ridgereg = Ridge(alpha=10, solver='saga', fit_intercept= True)
```

```python
ridgereg.fit(X_train,y_train)
```

```python
X_train
```

```python
y_pred = ridgereg.predict(X_test)
```

```python
y_pred
```

```python
y_pred=TargetVarScalerFit.inverse_transform(y_pred)
```

```python
y_pred = y_pred.reshape(-1)
```

```python
y_pred
```

```python
y_test_orig=TargetVarScalerFit.inverse_transform(y_test)
```

```python
y_test_orig
```

```python
y_test_orig = y_test_orig.reshape(-1)
```

```python
y_test_orig
```

```python
APE=100*(abs(y_test_orig-y_pred)/y_test_orig)
```

```python
print('The Accuracy of Ridge model is:', np.mean(APE))
```

```python
MAE=(abs(y_test_orig-y_pred))
print('The Accuracy of Ridge model is:',np.mean(MAE))
```

```python
import statistics
var = (statistics.variance(y_pred))
print(var)
chi_sq = np.sum(((y_test_orig-y_pred)**2)/var)
red_chi_sq = chi_sq/5074
print('The reduced chi squared value for Ridge is', red_chi_sq)
```
