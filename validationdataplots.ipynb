```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import csv
```

```python
df = pd.read_csv('150Cvalidationdata.csv')
```

```python
df
```

```python
x = df['True Plastic Strain'].to_numpy()
```

```python
y1 = df['True Stress'].to_numpy()
y2 = df['XGB Predicted Stress'].to_numpy()
```

```python
x
```

```python
y1
```

```python
y2
```

```python
plt.title("True Stress vs XGB Predicted Stress")
plt.xlabel("Strain")
plt.ylabel("Stress")
plt.ylim(0, 1000)
plt.xlim(0, 0.1)
plt.plot(x, y1, color ="red", label = 'True Stress')
plt.plot(x, y2, color ="blue", label = 'XGB Predicted Stress')
plt.legend()
plt.savefig('xgbprediction.png',dpi=300)
plt.show()
```

```python
import statistics
var = (statistics.variance(y2))
print(var)
chi_sq = np.sum(((y2-y1)**2)/var)
red_chi_sq = chi_sq/5074
print('The reduced chi squared value for XGB is', red_chi_sq)
```
